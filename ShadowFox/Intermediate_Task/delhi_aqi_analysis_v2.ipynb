```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../datasets/delhiaqi.csv')

print(df.head())
```

```python
df['date'] = pd.to_datetime(df['date'])

print(df.shape)
print(df.describe())
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['pm2_5'])
plt.title('PM2.5 Trend Over Time')
plt.xlabel('Date')
plt.ylabel('PM2.5')
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['pm10'])
plt.title('PM10 Trend Over Time')
plt.xlabel('Date')
plt.ylabel('PM10')
plt.show()
```

```python
pollutants = ['co','no','no2','o3','so2','pm2_5','pm10','nh3']

avg_values = df[pollutants].mean()

plt.figure(figsize=(10,5))
avg_values.plot(kind='bar')
plt.title('Average Pollutant Levels')
plt.ylabel('Average Value')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['co','no','no2','o3','so2','pm2_5','pm10','nh3']].corr(),
            annot=True,
            cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['pm2_5'], bins=20)
plt.title('Distribution of PM2.5')
plt.xlabel('PM2.5')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['pm2_5'], df['pm10'])
plt.title('PM2.5 vs PM10')
plt.xlabel('PM2.5')
plt.ylabel('PM10')
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df[['co','no','no2','o3','so2','pm2_5','pm10','nh3']]
)

plt.title('Pollutant Distribution')
plt.xticks(rotation=45)
plt.show()
```
