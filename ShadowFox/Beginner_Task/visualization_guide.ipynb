## 📈 SECTION 1: Line Plot
**What it shows:** How a value changes over time or a sequence.  
**Use case:** Monthly temperature, stock prices, exam scores over time.

```python
# ============================================================
# MATPLOTLIB LINE PLOT
# Average monthly temperature tracking
# ============================================================

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperature = [15, 17, 21, 26, 31, 35, 34, 33, 29, 24, 19, 16]

plt.figure(figsize=(10, 5))

plt.plot(months, temperature,
         color='royalblue',
         linewidth=2,
         marker='o',
         markersize=8,
         markerfacecolor='red',
         label='Temperature (°C)')

plt.axhline(y=25, color='orange', linestyle='--', alpha=0.7, label='25°C Reference')
plt.title('Average Monthly Temperature — Matplotlib', fontsize=15, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Matplotlib Line Plot: Shows temperature peaking in June (35°C) and lowest in January (15°C)')
```

```python
# ============================================================
# SEABORN LINE PLOT
# Same data — notice the cleaner code using DataFrame!
# ============================================================

df_temp = pd.DataFrame({
    'Month': months,
    'Temperature': temperature
})

plt.figure(figsize=(10, 5))

sns.lineplot(data=df_temp, x='Month', y='Temperature',
             marker='o', color='steelblue', linewidth=2.5,
             markersize=9, label='Temperature (°C)')

plt.axhline(y=25, color='coral', linestyle='--', alpha=0.8, label='25°C Reference')
plt.title('Average Monthly Temperature — Seaborn Style', fontsize=15, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Seaborn Line Plot: Same result with column names instead of raw arrays!')
```

## 📊 SECTION 2: Bar Chart
**What it shows:** Comparison across different categories.  
**Use case:** Sales by month, population by city, votes by candidate.

```python
# ============================================================
# MATPLOTLIB BAR CHART
# Monthly book sales comparison
# ============================================================

months_bar = ['January', 'February', 'March', 'April', 'May', 'June']
books_sold = [120, 95, 140, 180, 160, 200]
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7', '#DDA0DD']

plt.figure(figsize=(10, 6))
bars = plt.bar(months_bar, books_sold, color=colors, edgecolor='black', linewidth=0.8)

for bar, value in zip(bars, books_sold):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 3,
             str(value), ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.title('Monthly Book Sales — First Half of 2024 (Matplotlib)', fontsize=15, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Books Sold', fontsize=12)
plt.ylim(0, 230)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Matplotlib Bar Chart: June has the highest sales (200 books)!')
```

```python
# ============================================================
# SEABORN BAR CHART
# Same data with beautiful built-in palette
# ============================================================

df_sales = pd.DataFrame({
    'Month': months_bar,
    'Books_Sold': books_sold
})

plt.figure(figsize=(10, 6))
sns.barplot(data=df_sales, x='Month', y='Books_Sold',
            palette='Set2', edgecolor='black', linewidth=0.8)

plt.title('Monthly Book Sales — Seaborn Style', fontsize=15, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Books Sold', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Seaborn Bar Chart: Beautiful Set2 palette with half the code!')
```

## 🔵 SECTION 3: Scatter Plot
**What it shows:** Relationship (correlation) between two numeric variables.  
**Use case:** Study hours vs exam scores, house size vs price.

```python
# ============================================================
# MATPLOTLIB SCATTER PLOT
# Study hours vs exam scores for 50 students
# ============================================================

np.random.seed(42)
study_hours = np.random.uniform(1, 10, 50)
exam_scores = study_hours * 7 + np.random.normal(0, 8, 50)
exam_scores = np.clip(exam_scores, 20, 100)

plt.figure(figsize=(9, 6))
scatter = plt.scatter(study_hours, exam_scores,
                      c=exam_scores, cmap='RdYlGn',
                      s=80, edgecolors='black', linewidths=0.5, alpha=0.85)
plt.colorbar(scatter, label='Exam Score')

z = np.polyfit(study_hours, exam_scores, 1)
p = np.poly1d(z)
x_line = np.linspace(1, 10, 100)
plt.plot(x_line, p(x_line), 'r--', linewidth=2, label='Trend Line')

plt.title('Study Hours vs Exam Scores (Matplotlib)', fontsize=15, fontweight='bold')
plt.xlabel('Hours Studied per Day', fontsize=12)
plt.ylabel('Exam Score (out of 100)', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Matplotlib Scatter: Positive correlation visible — more study = higher scores!')
```

```python
# ============================================================
# SEABORN SCATTER PLOT
# regplot adds regression line AND confidence interval automatically!
# ============================================================

df_students = pd.DataFrame({
    'Study_Hours': study_hours,
    'Exam_Score': exam_scores
})

plt.figure(figsize=(9, 6))
sns.regplot(data=df_students, x='Study_Hours', y='Exam_Score',
            scatter_kws={'color': 'steelblue', 'edgecolors': 'black', 's': 80, 'alpha': 0.8},
            line_kws={'color': 'red', 'linewidth': 2, 'linestyle': '--'})

plt.title('Study Hours vs Exam Scores — Seaborn with Regression', fontsize=15, fontweight='bold')
plt.xlabel('Hours Studied per Day', fontsize=12)
plt.ylabel('Exam Score (out of 100)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Seaborn Scatter: Notice the shaded confidence interval — added automatically!')
```

## 📉 SECTION 4: Histogram
**What it shows:** Distribution (spread/shape) of a single numeric variable.  
**Use case:** Student heights, exam scores distribution, product prices.

```python
# ============================================================
# MATPLOTLIB HISTOGRAM
# Distribution of student heights
# ============================================================

np.random.seed(0)
heights = np.random.normal(loc=165, scale=10, size=200)

plt.figure(figsize=(10, 6))
n, bins, patches = plt.hist(heights, bins=20, color='steelblue',
                             edgecolor='white', linewidth=0.8, alpha=0.85)

bin_centers = 0.5 * (bins[:-1] + bins[1:])
for patch, center in zip(patches, bin_centers):
    if center < 155:
        patch.set_facecolor('#FF6B6B')
    elif center < 175:
        patch.set_facecolor('#4ECDC4')
    else:
        patch.set_facecolor('#45B7D1')

mean_height = np.mean(heights)
plt.axvline(mean_height, color='red', linestyle='--', linewidth=2,
            label=f'Mean: {mean_height:.1f} cm')

plt.title('Distribution of Student Heights (200 Students) — Matplotlib', fontsize=15, fontweight='bold')
plt.xlabel('Height (cm)', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f'📊 Matplotlib Histogram: Most students are between 155-175 cm. Mean = {mean_height:.1f} cm')
```

```python
# ============================================================
# SEABORN HISTOGRAM
# kde=True adds a smooth density curve on top automatically!
# ============================================================

df_heights = pd.DataFrame({'Height_cm': heights})

plt.figure(figsize=(10, 6))
sns.histplot(data=df_heights, x='Height_cm', bins=20,
             kde=True, color='steelblue', edgecolor='white', alpha=0.7,
             line_kws={'linewidth': 2.5, 'color': 'darkblue'})

plt.axvline(df_heights['Height_cm'].mean(), color='red', linestyle='--',
            linewidth=2, label=f'Mean: {heights.mean():.1f} cm')

plt.title('Distribution of Student Heights — Seaborn with KDE', fontsize=15, fontweight='bold')
plt.xlabel('Height (cm)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print('📊 Seaborn Histogram: The KDE curve shows the smooth bell-curve shape!')
```

## 🥧 SECTION 5: Pie Chart
**What it shows:** How a total is divided into parts (proportions).  
**Use case:** Market share, budget breakdown, survey results.

```python
# ============================================================
# MATPLOTLIB PIE CHART
# Favorite programming languages survey
# ============================================================

languages = ['Python', 'JavaScript', 'Java', 'C++', 'Others']
votes = [85, 50, 30, 20, 15]
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#C8A8E9']
explode = (0.08, 0, 0, 0, 0)

plt.figure(figsize=(9, 7))
wedges, texts, autotexts = plt.pie(
    votes, labels=languages, colors=colors,
    explode=explode, autopct='%1.1f%%',
    startangle=140, pctdistance=0.85, shadow=True
)

for autotext in autotexts:
    autotext.set_fontsize(11)
    autotext.set_fontweight('bold')

plt.title('Favorite Programming Languages\nSurvey of 200 Students', 
          fontsize=15, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print('📊 Matplotlib Pie Chart: Python dominates with 42.5% of votes!')
```

```python
# ============================================================
# SEABORN DONUT CHART (Modern alternative to pie chart)
# Seaborn doesn't have native pie chart — this uses Seaborn palette!
# ============================================================

colors_sns = sns.color_palette('pastel', len(languages))

fig, ax = plt.subplots(figsize=(9, 7))
wedges, texts, autotexts = ax.pie(
    votes, labels=languages, colors=colors_sns,
    autopct='%1.1f%%', startangle=140, pctdistance=0.80,
    wedgeprops=dict(width=0.6)
)

for autotext in autotexts:
    autotext.set_fontsize(11)
    autotext.set_fontweight('bold')

ax.text(0, 0, 'Languages\nSurvey', ha='center', va='center',
        fontsize=13, fontweight='bold', color='#333333')

ax.set_title('Favorite Programming Languages — Seaborn Palette (Donut Style)',
             fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print('📊 Seaborn Donut Chart: Modern, clean look using Seaborn pastel palette!')
print('💡 Tip: For pie charts, Matplotlib is usually the better primary choice.')
```

## 🏆 SECTION 6: Combined Dashboard
The best of both worlds — Matplotlib layouts + Seaborn beauty!

```python
# ============================================================
# COMBINED MATPLOTLIB + SEABORN DASHBOARD
# Using Matplotlib for layout, Seaborn for chart style
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('📊 ShadowFox DataViz Dashboard — Matplotlib + Seaborn Combined',
             fontsize=16, fontweight='bold', y=1.02)

# --- Plot 1: Line Chart (top-left) ---
sns.lineplot(data=df_temp, x='Month', y='Temperature',
             marker='o', color='steelblue', linewidth=2, ax=axes[0, 0])
axes[0, 0].set_title('Monthly Temperature', fontweight='bold')
axes[0, 0].set_xlabel('Month')
axes[0, 0].set_ylabel('Temperature (°C)')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(True, alpha=0.3)

# --- Plot 2: Bar Chart (top-right) ---
sns.barplot(data=df_sales, x='Month', y='Books_Sold',
            palette='Set2', edgecolor='black', ax=axes[0, 1])
axes[0, 1].set_title('Monthly Book Sales', fontweight='bold')
axes[0, 1].set_xlabel('Month')
axes[0, 1].set_ylabel('Books Sold')
axes[0, 1].tick_params(axis='x', rotation=45)
axes[0, 1].grid(axis='y', alpha=0.3)

# --- Plot 3: Scatter Plot (bottom-left) ---
sns.scatterplot(data=df_students, x='Study_Hours', y='Exam_Score',
                color='coral', edgecolor='black', s=60, alpha=0.8, ax=axes[1, 0])
axes[1, 0].set_title('Study Hours vs Exam Scores', fontweight='bold')
axes[1, 0].set_xlabel('Study Hours')
axes[1, 0].set_ylabel('Exam Score')
axes[1, 0].grid(True, alpha=0.3)

# --- Plot 4: Histogram (bottom-right) ---
sns.histplot(data=df_heights, x='Height_cm', bins=20,
             kde=True, color='mediumpurple', edgecolor='white', ax=axes[1, 1])
axes[1, 1].set_title('Student Height Distribution', fontweight='bold')
axes[1, 1].set_xlabel('Height (cm)')
axes[1, 1].set_ylabel('Count')
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print('🎉 Dashboard complete! This shows the power of combining both libraries.')
```
